## Question 29

## Question 36
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
# Drama aur Action ka data alag kiya

df['Genre_Simple'] = df['Genre']
# sirf wohi rows jin me Drama ya Action ho
drama_gross = df[df['Genre'].str.contains('Drama', na=False)]['Gross']
action_gross = df[df['Genre'].str.contains('Action', na=False)]['Gross']

# Box plot ke liye data prepare
import pandas as pd
plot_df = pd.DataFrame({
    'Gross': pd.concat([drama_gross, action_gross]),
    'Genre_Type': ['Drama']*len(drama_gross) + ['Action']*len(action_gross)
})

plt.figure(figsize=(7,5))
sns.boxplot(x='Genre_Type', y='Gross', data=plot_df)
plt.title('Drama vs Action - Gross Comparison')
plt.show()
```

## Question 40
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = df.sort_values(by='IMDB_Rating', ascending=False).head(5)

# IMDB ko 100 scale par lana
top5['IMDB_100'] = top5['IMDB_Rating'] * 10

x = top5['Series_Title']
width = 0.35

plt.figure(figsize=(12,6))
plt.bar(x, top5['IMDB_100'], width=width, label='IMDB Rating (x10)', alpha=0.8)
plt.bar([p + width for p in range(len(x))], top5['Meta_score'], width=width, label='Meta Score', alpha=0.8)

# ye simple tarika hai student wala
plt.figure(figsize=(12,6))
labels = top5['Series_Title'].tolist()
imdb_vals = top5['IMDB_100'].tolist()
meta_vals = top5['Meta_score'].tolist()

import numpy as np
x_pos = np.arange(len(labels))
plt.bar(x_pos - 0.2, imdb_vals, 0.4, label='IMDB (scaled 100)')
plt.bar(x_pos + 0.2, meta_vals, 0.4, label='Meta Score')
plt.xticks(x_pos, labels, rotation=20, ha='right')
plt.title('Top 5 Movies - IMDB vs Meta Score Comparison')
plt.legend()
plt.tight_layout()
plt.show()
```
